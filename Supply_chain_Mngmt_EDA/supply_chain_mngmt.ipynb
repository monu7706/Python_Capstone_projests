## 🔍 **1. Project Overview**

Company X ships thousands of orders daily using multiple courier partners.  
Courier charges depend on:

-  Total shipment weight  
-  Delivery zone (based on pincode distance)

Company X wants to verify:

- Whether courier companies charged correctly  
- Whether any order is **overcharged** or **undercharged**

This project uses **6 datasets** to calculate expected charges and compare them with billed charges.

---

##  **2. Datasets Used**

| File Name | Description |
|----------|-------------|
| Company X - Order Report.xlsx | Order ID, SKU, Quantity |
| Company X - SKU Master.xlsx | SKU-wise product weight |
| Company X - Pincode Zones.xlsx | Warehouse → Customer pincode → Zone |
| Courier Company - Invoice.xlsx | AWB, weight, zone, billed charges |
| Courier Company - Rates.xlsx | Forward & RTO rate card |
| Expected_Result.xlsx | Reference output format |
