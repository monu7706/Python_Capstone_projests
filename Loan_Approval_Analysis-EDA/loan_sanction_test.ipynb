### Step 6–7: Handling Missing Values
#### We identify missing values and fill them using median (for numeric) and mode (for categorical) to maintain data integrity.



```python
# Step 6: Check missing values
print("Missing values per column:")
print(df.isnull().sum())
```

```python
# Step 7: Fill missing values (if any)

# Numeric columns: fill with median
num_cols = ['ApplicantIncome', 'CoapplicantIncome', 'LoanAmount', 'Loan_Amount_Term', 'Credit_History']
for col in num_cols:
    df[col].fillna(df[col].median(), inplace=True)

# Categorical columns: fill with mode
cat_cols = ['Gender', 'Married', 'Dependents', 'Education', 'Self_Employed', 'Property_Area']
for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
# check again missing values
print(df.isnull().sum())
```

### TASK 2: DATA VISUALIZATION

### 2.1 UNIVARIATE ANALYSIS


#### A. Numeric Columns – Histograms

```python
# Numeric columns list
num_cols = ['ApplicantIncome', 'CoapplicantIncome', 'LoanAmount', 'Loan_Amount_Term', 'Credit_History']

plt.figure(figsize=(15, 10))

for i, col in enumerate(num_cols, 1):
    plt.subplot(3, 3, i)
    plt.hist(df[col], bins=20, color='skyblue', edgecolor='black')
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

### ### Histograms for Numeric Columns
#### Histograms help us understand the distribution of numeric variables such as income, loan amount, and loan term.  
#### We can identify skewness, spread, and concentration of values.


#### B. Numeric Columns – Box Plots (Outliers Check)


```python
plt.figure(figsize=(15, 10))

for i, col in enumerate(num_cols, 1):
    plt.subplot(3, 3, i)
    sns.boxplot(y=df[col], color='lightgreen')
    plt.title(f'Boxplot of {col}')

plt.tight_layout()
plt.show()
```

### ### Box Plots for Outlier Detection
#### Box plots help identify outliers and understand the spread of numeric variables.


#### C. Categorical Columns – Bar Charts

```python
cat_cols = ['Gender', 'Married', 'Dependents', 'Education', 'Self_Employed', 'Property_Area']

for col in cat_cols:
    plt.figure(figsize=(6,4))
    df[col].value_counts().plot(kind='bar', color='coral', edgecolor='black')
    plt.title(f'Count of categories in {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=45)
    plt.show()
```

### ### Bar Charts for Categorical Variables
#### Bar charts show how many applicants fall into each category (e.g., Male/Female, Graduate/Not Graduate).


#### D. Categorical Columns – Pie Charts


```python
for col in cat_cols:
    if df[col].nunique() <= 5:  # Only small category columns
        plt.figure(figsize=(5,5))
        df[col].value_counts().plot(kind='pie', autopct='%1.1f%%', startangle=90)
        plt.title(f'Pie chart of {col}')
        plt.ylabel('')
        plt.show()
```

### ### Pie Charts for Categorical Variables
#### Pie charts show the percentage distribution of categories, useful for variables with fewer unique values.

###  B. Stacked Bar Chart (Categorical vs Categorical)


#### Example: Property_Area vs Married

```python
crosstab = pd.crosstab(df['Property_Area'], df['Married'])

crosstab.plot(kind='bar', stacked=True, figsize=(7,5), colormap='viridis')
plt.title('Marital Status by Property Area')
plt.xlabel('Property Area')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```
